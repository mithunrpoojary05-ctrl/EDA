```python
df["Date of Admission"]=pd.to_datetime(df["Date of Admission"])
```

```python
df["Discharge Date"]=pd.to_datetime(df["Discharge Date"])
```

```python
df_pivot=pd.pivot_table(df,
   index="Medical Condition",
   columns="Insurance Provider",
   values="Billing Amount",
   aggfunc="sum")
print(df_pivot)
```

```python
df_pivot.plot(kind="bar",stacked=True,figsize=(12,5))
plt.title("plots comparing billing amount across medical conditions and insurence providers")
plt.xlabel("Medical Condition")
plt.ylabel("Billing Amount")
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
sns.violinplot(x="Insurance Provider",y="Billing Amount",data=df)
plt.title("Billing amount by insurance provider")
plt.ylabel("Billing Amount")
plt.xlabel("Insurance Provider")
plt.xticks(rotation=55)
plt.show()
```

```python
sns.lineplot(x="Date of Admission",y="Discharge Date",data=df)
plt.title("HHH")
plt.ylabel("Discharge Date")
plt.xlabel("Date of Admission")
plt.show()
```
