```python
parental_avg=df.groupby("parental level of education")[
    ["math score", "reading score", "writing score"]
].mean()

print("\nAverage scores by test preparation:")
print(parental_avg)
```

```python
parental_avg.plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Average Test Scores by Parental Education Level")
plt.xlabel("Parental Education Level")
plt.ylabel("Average Score")
plt.xticks(rotation=30)
plt.legend(title="Subject")
plt.tight_layout()
plt.show()
```

```python
lunch_avg=df.groupby("parental level of education")[
    ["math score", "reading score", "writing score"]].mean()

print("\nAverage scores by test preparation:")
print(lunch_avg)
```
