```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.histplot(df["Sales"],bins=8)
plt.title("distribution of Sales")
plt.show()
```

```python
sns.histplot(df["Profit"],bins=8)
plt.title("distribution of Profit")
plt.show()
```

```python
sns.boxplot(x="Profit",y="Sales",data=df)
plt.show()
```

```python
sns.scatterplot(x="Profit",y="Sales",data=df)
plt.show()
```

```python
sns.barplot(x="Profit",y="Sales",data=df)
plt.show()
```

```python
numeric_col=["Unit_Price", "Quantity", "Discount_Percent", "Sales", "Profit", "Profit_Margin_Percent", "Cost"]
```

```python
correlation=df[numeric_col].corr()
```

```python
correlation
```

```python
sns.heatmap(correlation,
            annot=True,      
            cmap="Greens")
plt.show()
```
